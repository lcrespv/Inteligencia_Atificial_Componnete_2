```python
import zipfile
from collections import Counter

ruta = "/content/fruits360/fruits.zip"

with zipfile.ZipFile(ruta, "r") as zip_ref:
    archivos = zip_ref.namelist()

# Identificar las carpetas principales
carpetas = Counter()

for archivo in archivos:
    partes = archivo.split("/")

    if len(partes) >= 3:
        carpetas["/".join(partes[:3])] += 1

print("=== CARPETAS PRINCIPALES ===")
for carpeta, cantidad in carpetas.most_common():
    print(f"{carpeta}: {cantidad} archivos")

# Identificar las imágenes del dataset 100x100
imagenes = [
    archivo for archivo in archivos
    if archivo.lower().endswith((".jpg", ".jpeg", ".png"))
    and archivo.startswith("fruits-360_100x100/fruits-360/")
]

print("\n=== TOTAL DE IMÁGENES 100x100 ===")
print("Imágenes encontradas:", len(imagenes))

# Contar las imágenes por conjunto y clase
conteo = Counter()
clases = set()

for archivo in imagenes:
    partes = archivo.split("/")

    if len(partes) >= 5:
        conjunto = partes[2]
        clase = partes[3]

        if conjunto in ("Training", "Test", "Validation"):
            conteo[conjunto] += 1
            clases.add(clase)

print("\n=== DISTRIBUCIÓN ORIGINAL ===")
for conjunto, cantidad in conteo.items():
    print(f"{conjunto}: {cantidad} imágenes")

print("\n=== CLASES ===")
print("Número de clases:", len(clases))

print("\nPrimeras 20 clases:")
for clase in sorted(clases)[:20]:
    print("-", clase)
```

```python

import os
import glob
import zipfile
import io
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image
from collections import Counter

# Buscar el ZIP descargado
archivos_zip = glob.glob("/content/fruits360/*.zip")

if not archivos_zip:
    raise FileNotFoundError("No se encontró Fruits-360.")

ruta = next(
    (p for p in archivos_zip if os.path.basename(p) == "fruits.zip"),
    archivos_zip[0]
)

clases = [
    "Apple Red 1",
    "Banana 1",
    "Orange 1",
    "Strawberry 1"
]

base = "fruits-360_100x100/fruits-360/"

# Identificar las imágenes seleccionadas
registros = []

with zipfile.ZipFile(ruta, "r") as z:
    for nombre in z.namelist():
        partes = nombre.split("/")

        if (
            len(partes) == 5
            and nombre.startswith(base)
            and partes[2] in ("Training", "Test")
            and partes[3] in clases
            and partes[4].lower().endswith((".jpg", ".jpeg", ".png"))
        ):
            registros.append({
                "ruta": nombre,
                "conjunto": partes[2],
                "clase": partes[3]
            })

df = pd.DataFrame(registros)

print("ZIP encontrado:", ruta)
print("Imágenes seleccionadas:", len(df))
print("\nDistribución original:")
print(pd.crosstab(df["clase"], df["conjunto"]))
```

```python

from PIL import Image
from collections import Counter
import zipfile
import io

archivos_danados = []
archivos_vacios = []
dimensiones = Counter()
modos_color = Counter()

with zipfile.ZipFile(ruta, "r") as z:
    for nombre in df["ruta"]:
        try:
            with z.open(nombre) as archivo:
                datos = archivo.read()

            if len(datos) == 0:
                archivos_vacios.append(nombre)
                continue

            with Image.open(io.BytesIO(datos)) as img:
                img.load()
                dimensiones[img.size] += 1
                modos_color[img.mode] += 1

        except Exception as error:
            archivos_danados.append((nombre, str(error)))

dimensiones_incorrectas = sum(
    cantidad
    for tamano, cantidad in dimensiones.items()
    if tamano != (100, 100)
)

print("RESULTADOS DE INTEGRIDAD")
print("Total revisado:", len(df))
print("Archivos vacíos:", len(archivos_vacios))
print("Archivos dañados:", len(archivos_danados))
print("Dimensiones incorrectas:", dimensiones_incorrectas)
print("Dimensiones encontradas:", dict(dimensiones))
print("Modos de color:", dict(modos_color))

if archivos_danados:
    print("\nEjemplos de errores:")
    for archivo in archivos_danados[:5]:
        print(archivo)
```

```python

import cv2
import numpy as np
import pandas as pd
import zipfile
import io
import matplotlib.pyplot as plt
from PIL import Image

resultados = []

with zipfile.ZipFile(ruta, "r") as z:
    for _, fila in df.iterrows():
        nombre = fila["ruta"]

        try:
            with z.open(nombre) as archivo:
                datos = archivo.read()

            with Image.open(io.BytesIO(datos)) as imagen:
                img = np.array(imagen.convert("RGB"))

            gris = cv2.cvtColor(
                img, cv2.COLOR_RGB2GRAY
            )

            nitidez = cv2.Laplacian(
                gris, cv2.CV_64F
            ).var()

            resultados.append({
                "ruta": nombre,
                "clase": fila["clase"],
                "nitidez": nitidez
            })

        except Exception:
            continue

df_calidad = pd.DataFrame(resultados)
df_calidad = df_calidad.sort_values("nitidez")

print("Imágenes analizadas:", len(df_calidad))
print("\nResumen de nitidez:")
print(df_calidad["nitidez"].describe())

print("\n10 imágenes con menor nitidez:")
print(
    df_calidad[["clase", "nitidez"]]
    .head(10)
    .to_string(index=False)
)

# Mostrar las 12 imágenes con menor nitidez
seleccion = df_calidad.head(12)

fig, axes = plt.subplots(3, 4, figsize=(11, 9))

with zipfile.ZipFile(ruta, "r") as z:
    for ax, (_, fila) in zip(axes.flat, seleccion.iterrows()):
        with z.open(fila["ruta"]) as archivo:
            datos = archivo.read()

        with Image.open(io.BytesIO(datos)) as imagen:
            img = imagen.convert("RGB").copy()

        ax.imshow(img)
        ax.set_title(
            f"{fila['clase']}\n"
            f"Valor: {fila['nitidez']:.1f}",
            fontsize=9
        )
        ax.axis("off")

plt.suptitle("Imágenes candidatas a revisión de nitidez")
plt.tight_layout()
plt.savefig("/content/figura4_nitidez.png", dpi=300)
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Datos del conjunto de entrenamiento (Train)
datos = {
    "Clase": [
        "Apple Red 1",
        "Banana 1",
        "Orange 1",
        "Strawberry 1"
    ],
    "Cantidad": [393, 392, 383, 394]
}

df_balance = pd.DataFrame(datos)

# Calcular porcentajes
total = df_balance["Cantidad"].sum()
df_balance["Porcentaje"] = (
    df_balance["Cantidad"] / total * 100
)

# Calcular el desbalance
maximo = df_balance["Cantidad"].max()
minimo = df_balance["Cantidad"].min()
diferencia = maximo - minimo
relacion = maximo / minimo

# Mostrar resultados
print("ANÁLISIS DE DESBALANCE DE CLASES")
print("-" * 45)
print(df_balance.to_string(index=False))
print("\nTotal de imágenes:", total)
print("Diferencia máxima:", diferencia)
print("Relación de desbalance:", round(relacion, 3))

# Crear gráfica
plt.figure(figsize=(10, 6))

colores = ["#E74C3C", "#F4D03F", "#F39C12", "#E91E63"]

barras = plt.bar(
    df_balance["Clase"],
    df_balance["Cantidad"],
    color=colores,
    edgecolor="black"
)

# Mostrar cantidad y porcentaje encima de cada barra
for barra, cantidad, porcentaje in zip(
    barras,
    df_balance["Cantidad"],
    df_balance["Porcentaje"]
):
    plt.text(
        barra.get_x() + barra.get_width() / 2,
        barra.get_height() + 5,
        f"{cantidad} ({porcentaje:.2f}%)",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

plt.title(
    "Análisis de desbalance de las cuatro clases - Fruits-360",
    fontsize=14,
    fontweight="bold"
)
plt.xlabel("Clases de frutas")
plt.ylabel("Número de imágenes de entrenamiento")
plt.ylim(0, 450)
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()

# Guardar la gráfica
plt.savefig(
    "/content/analisis_desbalance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

import os
import zipfile
import shutil
from sklearn.model_selection import train_test_split
from collections import Counter

# Ruta del dataset original
ruta_zip = "/content/fruits360/fruits.zip"

# Carpeta donde guardaremos las imágenes preparadas
destino = "/content/fruits360_preprocesado"

# Cuatro clases seleccionadas
clases = [
    "Apple Red 1",
    "Banana 1",
    "Orange 1",
    "Strawberry 1"
]

# Verificar que existe el ZIP
if not os.path.exists(ruta_zip):
    raise FileNotFoundError("No se encontró el archivo Fruits-360.")

# Crear carpeta limpia para esta ejecución
if os.path.exists(destino):
    shutil.rmtree(destino)

for conjunto in ["Train", "Validation", "Test"]:
    for clase in clases:
        os.makedirs(
            os.path.join(destino, conjunto, clase),
            exist_ok=True
        )

# Identificar imágenes dentro del ZIP
imagenes_train = []
etiquetas_train = []
imagenes_test = []

with zipfile.ZipFile(ruta_zip, "r") as z:
    for nombre in z.namelist():
        partes = nombre.split("/")

        if (
            len(partes) == 5
            and partes[0] == "fruits-360_100x100"
            and partes[1] == "fruits-360"
            and partes[3] in clases
            and partes[4].lower().endswith(
                (".jpg", ".jpeg", ".png")
            )
        ):
            if partes[2] == "Training":
                imagenes_train.append(nombre)
                etiquetas_train.append(partes[3])

            elif partes[2] == "Test":
                imagenes_test.append(nombre)

# Dividir el Training original en 80 % y 20 %
train, validation = train_test_split(
    imagenes_train,
    test_size=0.20,
    random_state=42,
    stratify=etiquetas_train
)

# Extraer las imágenes a las carpetas correspondientes
with zipfile.ZipFile(ruta_zip, "r") as z:

    for conjunto, lista in [
        ("Train", train),
        ("Validation", validation),
        ("Test", imagenes_test)
    ]:
        for nombre in lista:
            clase = nombre.split("/")[3]
            archivo = os.path.basename(nombre)

            ruta_salida = os.path.join(
                destino, conjunto, clase, archivo
            )

            with z.open(nombre) as origen:
                with open(ruta_salida, "wb") as salida:
                    shutil.copyfileobj(origen, salida)

# Mostrar resultados
print("ORGANIZACIÓN DEL DATASET COMPLETADA")
print("-" * 45)

for conjunto in ["Train", "Validation", "Test"]:
    print(f"\n{conjunto}:")

    total = 0

    for clase in clases:
        carpeta = os.path.join(
            destino, conjunto, clase
        )

        cantidad = len(os.listdir(carpeta))
        total += cantidad

        print(f"  {clase}: {cantidad}")

    print(f"  TOTAL: {total}")

print("\nCarpeta creada:", destino)
```

```python

import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from PIL import Image

ruta_train = "/content/fruits360_preprocesado/Train"

clases = [
    "Apple Red 1",
    "Banana 1",
    "Orange 1",
    "Strawberry 1"
]

# Transformaciones suaves con fondo blanco
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomRotation(
        0.04,
        fill_mode="constant",
        fill_value=255.0
    ),
    tf.keras.layers.RandomTranslation(
        0.03, 0.03,
        fill_mode="constant",
        fill_value=255.0
    ),
    tf.keras.layers.RandomZoom(
        height_factor=(-0.08, 0.0),
        width_factor=(-0.08, 0.0),
        fill_mode="constant",
        fill_value=255.0
    )
])

fig, axes = plt.subplots(4, 4, figsize=(12, 12))

for i, clase in enumerate(clases):

    carpeta = os.path.join(ruta_train, clase)

    archivos = sorted([
        f for f in os.listdir(carpeta)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ])

    ruta = os.path.join(carpeta, archivos[0])

    imagen = np.array(
        Image.open(ruta).convert("RGB"),
        dtype=np.float32
    )

    axes[i, 0].imshow(imagen.astype(np.uint8))
    axes[i, 0].set_title(f"{clase}\nOriginal")
    axes[i, 0].axis("off")

    for j in range(1, 4):

        transformada = data_augmentation(
            np.expand_dims(imagen, axis=0),
            training=True
        )[0].numpy()

        transformada = np.clip(
            transformada, 0, 255
        ).astype(np.uint8)

        axes[i, j].imshow(transformada)
        axes[i, j].set_title(f"Transformación {j}")
        axes[i, j].axis("off")

plt.suptitle(
    "Data augmentation corregido - Fruits-360",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    "/content/figura_data_augmentation_corregido.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Transformaciones generadas correctamente.")
print("Total de ejemplos visualizados: 16")
```

```python

import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from PIL import Image

ruta_train = "/content/fruits360_preprocesado/Train"

clases = [
    "Apple Red 1",
    "Banana 1",
    "Orange 1",
    "Strawberry 1"
]

# Reproducibilidad de las transformaciones
tf.keras.utils.set_random_seed(42)

# Transformaciones moderadas
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomRotation(
        0.04,
        fill_mode="constant",
        fill_value=255.0
    ),
    tf.keras.layers.RandomTranslation(
        0.03, 0.03,
        fill_mode="constant",
        fill_value=255.0
    ),
    tf.keras.layers.RandomZoom(
        height_factor=(-0.04, 0.0),
        width_factor=(-0.04, 0.0),
        fill_mode="constant",
        fill_value=255.0
    )
])

# Añadir margen blanco sin cambiar el tamaño final
def agregar_margen(imagen, escala=0.82):

    alto, ancho = imagen.shape[:2]

    nuevo_ancho = int(ancho * escala)
    nuevo_alto = int(alto * escala)

    imagen_pil = Image.fromarray(imagen)

    reducida = imagen_pil.resize(
        (nuevo_ancho, nuevo_alto),
        Image.Resampling.BILINEAR
    )

    fondo = Image.new(
        "RGB",
        (ancho, alto),
        (255, 255, 255)
    )

    x = (ancho - nuevo_ancho) // 2
    y = (alto - nuevo_alto) // 2

    fondo.paste(reducida, (x, y))

    return np.array(fondo, dtype=np.float32)


# Visualización de las cuatro clases
fig, axes = plt.subplots(4, 5, figsize=(15, 12))

for i, clase in enumerate(clases):

    carpeta = os.path.join(ruta_train, clase)

    archivos = sorted([
        f for f in os.listdir(carpeta)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ])

    ruta = os.path.join(carpeta, archivos[0])

    original = np.array(
        Image.open(ruta).convert("RGB")
    )

    # Crear versión con margen blanco
    con_margen = agregar_margen(original)

    # Imagen original
    axes[i, 0].imshow(original)
    axes[i, 0].set_title(f"{clase}\nOriginal")
    axes[i, 0].axis("off")

    # Imagen con margen
    axes[i, 1].imshow(
        con_margen.astype(np.uint8)
    )
    axes[i, 1].set_title("Con margen")
    axes[i, 1].axis("off")

    # Tres transformaciones
    for j in range(2, 5):

        transformada = data_augmentation(
            np.expand_dims(con_margen, axis=0),
            training=True
        )[0].numpy()

        transformada = np.clip(
            transformada, 0, 255
        ).astype(np.uint8)

        axes[i, j].imshow(transformada)
        axes[i, j].set_title(
            f"Transformación {j-1}"
        )
        axes[i, j].axis("off")

plt.suptitle(
    "Data augmentation con margen de seguridad - Fruits-360",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    "/content/figura_augmentation_final.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Proceso completado.")
print("Clases procesadas:", len(clases))
print("Ejemplos visualizados:", 20)
print("Dimensiones finales: 100 x 100 x 3")
print("Archivos originales: sin modificaciones")
```

```python

import os
import numpy as np
import pandas as pd
import tensorflow as tf
from PIL import Image

# ==========================================
# 1. CONFIGURACIÓN GENERAL
# ==========================================

ruta_dataset = "/content/fruits360_preprocesado"

clases = [
    "Apple Red 1",
    "Banana 1",
    "Orange 1",
    "Strawberry 1"
]

codificacion = {
    clase: indice
    for indice, clase in enumerate(clases)
}

conjuntos = ["Train", "Validation", "Test"]

cantidades_esperadas = {
    "Train": 1562,
    "Validation": 391,
    "Test": 654
}

configuraciones = {
    "CNN propia": (100, 100),
    "MobileNetV2": (224, 224),
    "EfficientNetB0": (224, 224),
    "ConvNeXtTiny": (224, 224)
}

# ==========================================
# 2. CARGAR RUTAS Y ETIQUETAS
# ==========================================

registros = []

for conjunto in conjuntos:
    for clase in clases:

        carpeta = os.path.join(
            ruta_dataset, conjunto, clase
        )

        if not os.path.isdir(carpeta):
            raise FileNotFoundError(carpeta)

        archivos = sorted([
            archivo
            for archivo in os.listdir(carpeta)
            if archivo.lower().endswith(
                (".jpg", ".jpeg", ".png")
            )
        ])

        for archivo in archivos:
            registros.append({
                "Conjunto": conjunto,
                "Clase": clase,
                "Etiqueta": codificacion[clase],
                "Ruta": os.path.join(
                    carpeta, archivo
                )
            })

df = pd.DataFrame(registros)

print("1. VERIFICACIÓN DE CONJUNTOS")
print("-" * 50)

for conjunto in conjuntos:

    cantidad = len(
        df[df["Conjunto"] == conjunto]
    )

    esperado = cantidades_esperadas[conjunto]

    print(
        f"{conjunto}: {cantidad} imágenes "
        f"(esperadas: {esperado})"
    )

    assert cantidad == esperado, (
        f"Cantidad incorrecta en {conjunto}"
    )

assert len(df) == 2607

print("\nTotal general:", len(df))

# ==========================================
# 3. VERIFICAR ETIQUETAS
# ==========================================

print("\n2. VERIFICACIÓN DE ETIQUETAS")
print("-" * 50)

for clase, numero in codificacion.items():
    print(f"{clase} -> {numero}")

assert set(df["Etiqueta"]) == {0, 1, 2, 3}

assert df["Etiqueta"].notna().all()

print("Etiquetas verificadas correctamente.")

# ==========================================
# 4. VERIFICAR SEPARACIÓN DE CONJUNTOS
# ==========================================

print("\n3. VERIFICACIÓN DE SEPARACIÓN")
print("-" * 50)

rutas_train = set(
    df[df["Conjunto"] == "Train"]["Ruta"]
)

rutas_val = set(
    df[df["Conjunto"] == "Validation"]["Ruta"]
)

rutas_test = set(
    df[df["Conjunto"] == "Test"]["Ruta"]
)

assert rutas_train.isdisjoint(rutas_val)
assert rutas_train.isdisjoint(rutas_test)
assert rutas_val.isdisjoint(rutas_test)

print("No existen rutas compartidas entre conjuntos.")

# ==========================================
# 5. PREPARAR ENTRADAS PARA LOS MODELOS
# ==========================================

print("\n4. PREPARACIÓN POR ARQUITECTURA")
print("-" * 50)

# Usaremos una imagen de cada clase
# para verificar las cuatro arquitecturas.

muestras = []

for clase in clases:

    fila = df[
        (df["Conjunto"] == "Train")
        & (df["Clase"] == clase)
    ].iloc[0]

    muestras.append(
        (clase, fila["Ruta"], fila["Etiqueta"])
    )

def preparar_imagen(ruta, modelo):

    with Image.open(ruta) as imagen:
        imagen = imagen.convert("RGB")

        tamano = configuraciones[modelo]

        imagen = imagen.resize(
            (tamano[1], tamano[0]),
            Image.Resampling.BILINEAR
        )

        arreglo = np.array(
            imagen, dtype=np.float32
        )

    if modelo == "CNN propia":

        arreglo = arreglo / 255.0

    elif modelo == "MobileNetV2":

        arreglo = (
            tf.keras.applications.mobilenet_v2
            .preprocess_input(arreglo)
        )

    # EfficientNetB0 y ConvNeXtTiny:
    # Mantener valores entre 0 y 255.
    # Se utilizará su preprocesamiento
    # interno al construir los modelos.

    return arreglo

resultados = []

for modelo in configuraciones:

    print(f"\nMODELO: {modelo}")

    for clase, ruta, etiqueta in muestras:

        imagen = preparar_imagen(
            ruta, modelo
        )

        dimensiones = imagen.shape

        minimo = float(imagen.min())
        maximo = float(imagen.max())

        assert dimensiones == (
            *configuraciones[modelo], 3
        )

        assert imagen.dtype == np.float32

        assert np.isfinite(imagen).all()

        if modelo == "CNN propia":
            assert minimo >= 0
            assert maximo <= 1

        elif modelo == "MobileNetV2":
            assert minimo >= -1
            assert maximo <= 1

        else:
            assert minimo >= 0
            assert maximo <= 255

        resultados.append({
            "Modelo": modelo,
            "Clase": clase,
            "Etiqueta": etiqueta,
            "Dimensiones": str(dimensiones),
            "Mínimo": round(minimo, 4),
            "Máximo": round(maximo, 4),
            "Estado": "Correcto"
        })

    print("4 clases verificadas correctamente.")

# ==========================================
# 6. MOSTRAR TABLA DE RESULTADOS
# ==========================================

tabla_resultados = pd.DataFrame(resultados)

print("\n5. TABLA DE VERIFICACIÓN")
print("-" * 50)

print(
    tabla_resultados.to_string(index=False)
)

# Guardar evidencia en CSV
ruta_csv = "/content/verificacion_preprocesamiento.csv"

tabla_resultados.to_csv(
    ruta_csv,
    index=False,
    encoding="utf-8-sig"
)

print("\nRESULTADO FINAL")
print("=" * 50)

print("Conjuntos verificados: 3")
print("Clases verificadas: 4")
print("Arquitecturas verificadas: 4")
print("Total de imágenes registradas:", len(df))
print("Muestras de entrada comprobadas:", len(resultados))

print("\nVerificación de muestras completada.")
print("No se ha entrenado ningún modelo.")
print("Archivo de evidencia:", ruta_csv)
```
